```python
## unique characters in dataset, later used as vocab 
chars = sorted(list(set(text)))
vocab_size = len(chars)
print(''.join(chars))
print(vocab_size)
```

```python
## tokenizing Text (create a mapping from string to integers)

stoi = {ch:i for i,ch in enumerate(chars)}
itos = {i:ch for i,ch in enumerate(chars)}
encode = lambda s : [stoi[c] for c in s]  # str to int list
decode = lambda l : ''.join([itos[i] for i in l])  # int list to str

print(encode("hello"))
print(decode(encode("hello")))
```

```python
## tokenixe the original text and store it into a torch.Tensor

import torch
data = torch.tensor(encode(text),dtype=torch.long)
print(data.shape, data.dtype)
print(data[:100])   # corresponding tokenized text of first 1000 characters
```

```python
## split data into train and val set

n = int(0.9*len(data))  # 90%
train_data = data[:n]  # 90% data is train
val_data = data[n:]
```

```python
## for understanding how actually the language modeling/next word prediction works

block_size = 8    # context length 
x = train_data[:block_size]
y = train_data[1:block_size+1]

for t in range(block_size):
    context = x[:t+1]
    target = y[t]         # indexing a 1D PyTorch tensor returns a 0-dimensional scalar tensor
    print(f"when input is {context}, target is : {target}")
```

```python
## to use gpus efficiently, instead of providing a single training sample at a time, we provide batches of data where each sample is of block_size

torch.manual_seed(1337)
batch_size = 4
block_size = 8

def get_batch(split):
    data = train_data if split=='train' else val_data
    ix = torch.randint(len(data)-block_size , (batch_size,))   # generates random 4 positions to grab chunks from
    x = torch.stack([data[i:i+block_size] for i in ix])
    y = torch.stack([data[i+1:i+block_size+1] for i in ix])

    return x,y


xb , yb = get_batch('train')

print("inputs")
print(xb.shape)   # (B,T) ---- B = batch, T = time dimension/block/context
print(xb)
print('targets')
print(yb.shape)
print(yb)
print("******************************")
print()

for b in range(batch_size):
    for t in range(block_size):
        context = xb[b,:t+1]
        target = yb[b,t]
        print(f"when input is {context.tolist()}, target is : {target}")
```

```python
import torch
import torch.nn as nn
from torch.nn import functional as F
torch.manual_seed(1337)

class BigramLanguageModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()

        # each token directly reads of logits for next token from a lookup table

        self.token_embedding_table = nn.Embedding(vocab_size,vocab_size)

    def forward(self,idx,targets=None):
        ## idx and targets are both (B,T) shape tensors .... idx is input i.e x

        logits = self.token_embedding_table(idx)   # (B,T,C)  .... C = channel/vocab_size

        if targets is None:
            loss = None
        else:
            B,T,C = logits.shape
            logits = logits.view(B*T,C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits,targets)

        return logits,loss


# for prediction 

    def generate(self,idx,max_new_tokens):
        for _ in range(max_new_tokens):
            # get the predictions
            logits, loss = self(idx)
            # focus only on last time step
            logits = logits[:,-1,:]    # becomes (B,C) from (B,T,C)  #picks logits of last token of each batch
            # apply softmax to get prob
            probs = F.softmax(logits,dim=-1)   # (B,C)   # Apply softmax across the vocabulary dimension for each batch item individually
            # sample from the distribution
            idx_next = torch.multinomial(probs,num_samples=1)   # (B,1) -- Tells PyTorch to draw 1 token per batch item. 
            # append sampled index to the running seq
            idx = torch.cat((idx,idx_next), dim=1)    # (B,T+1)

        return idx

m = BigramLanguageModel(vocab_size)
logits,loss = m(xb,yb)
#print(logits)
print(logits.shape)
print(loss)
```

```python
## create a pytorch optimizer

optimizer = torch.optim.AdamW(m.parameters(),lr=1e-3)
```

```python
batch_size = 32
for steps in range(10000):
    # sample a batch of data
    xb,yb = get_batch('train')

    # evaluate the loss
    logits , loss = m(xb,yb)
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()
    print(loss.item())
```

```python
## self attention trick
torch.manual_seed(1337)
B,T,C = 4,8,2
x = torch.randn(B,T,C)
x.shape
```

```python
# Force float64 precision
x = torch.randn(B, T, C, dtype=torch.float64)

# Method 1: Loop
xbow = torch.zeros((B, T, C), dtype=torch.float64)
for b in range(B):
    for t in range(T):
        xprev = x[b, :t+1]
        xbow[b, t] = torch.mean(xprev, 0)
```

```python
# version 2  ; MATRIX MULT  ( dtype correction needed)
wei = torch.tril(torch.ones(T,T))
wei = wei/wei.sum(1,keepdim=True)
xbow2 = wei @ x
torch.allclose(xbow,xbow2)
```

```python
# Method 3: Matrix Multiplication  ( final version; we will use this)
tril = torch.tril(torch.ones(T, T, dtype=torch.float64))
wei = torch.zeros((T, T), dtype=torch.float64)
wei = wei.masked_fill(tril == 0, float('-inf'))
wei = F.softmax(wei, dim=-1)      # It turns raw numbers into weights that sum to 1.0 (100%)

xbow3 = wei @ x
```

```python
print("Are they close?", torch.allclose(xbow, xbow3))
```

```python
# the real self attention
torch.manual_seed(42)
B,T,C = 4,8,32
x = torch.randn(B, T, C)
head_size = 16
key = nn.Linear(C,head_size,bias=False)
query = nn.Linear(C,head_size,bias=False)
k = key(x)
q = query(x)

wei = q @ k.transpose(-2,-1)
tril = torch.tril(torch.ones(T, T))
# wei = torch.zeros((T, T), dtype=torch.float64)
wei = wei.masked_fill(tril == 0, float('-inf'))
wei = F.softmax(wei, dim=-1)

out = wei @ x
out.shape
```

```python
# we dont actually aggreagte x rather we create one other variable value in the same way as q and k 
# the real self attention
torch.manual_seed(42)
B,T,C = 4,8,32
x = torch.randn(B, T, C)
head_size = 16
key = nn.Linear(C,head_size,bias=False)
query = nn.Linear(C,head_size,bias=False)
value = nn.Linear(C,head_size,bias=False)
k = key(x)   # x says, here is what I have
q = query(x)  # x says, here is what I am interested in
v = value(x)  # x says, and here is my value if you find me interesting/ here is what I'll communicate if you find me interesting

wei = q @ k.transpose(-2,-1)
tril = torch.tril(torch.ones(T, T))
# wei = torch.zeros((T, T), dtype=torch.float64)
wei = wei.masked_fill(tril == 0, float('-inf'))
wei = F.softmax(wei, dim=-1)

out = wei @ v
out.shape
```

Notes:
- Attention is a **communication mechanism**. Can be seen as nodes in a directed graph looking at each other and aggregating information with a weighted sum from all nodes that point to them, with data-dependent weights.
- There is no notion of space. Attention simply acts over a set of vectors. This is why we need to positionally encode tokens.
- Each example across batch dimension is of course processed completely independently and never "talk" to each other
- In an "encoder" attention block just delete the single line that does masking with `tril`, allowing all tokens to communicate. This block here is called a "decoder" attention block because it has triangular masking, and is usually used in autoregressive settings, like language modeling.
- "self-attention" just means that the keys and values are produced from the same source as queries. In "cross-attention", the queries still get produced from x, but the keys and values come from some other, external source (e.g. an encoder module)
- "Scaled" attention additional divides `wei` by 1/sqrt(head_size). This makes it so when input Q,K are unit variance, wei will be unit variance too and Softmax will stay diffuse and not saturate too much. Illustration below

```python
# why layer norm?
# run below cells with and without layer norm to see the difference
k = torch.randn(B,T,head_size)
q = torch.randn(B,T,head_size)
wei = q @ k.transpose(-2, -1) * head_size**-0.5    # head_size**-0.5 --- this is layer norm, it is used to prevent the dot products from growing too large in magnitude, which can lead to vanishing gradients during training. By scaling the dot products by the inverse square root of the head size, we ensure that the values remain in a reasonable range, improving the stability and convergence of the model.
```
