```python
## read the dataset
with open('input.txt','r',encoding='utf-8') as f:
    text = f.read()
```

```python
## length of dataset in characters
print("length of text :",len(text))
```

```python
## let's see first 1000 characters
print(text[:1000])
```

```python
## unique characters in dataset, later used as vocab 
chars = sorted(list(set(text)))
vocab_size = len(chars)
print(''.join(chars))
print(vocab_size)
```

```python
## tokenizing Text (create a mapping from string to integers)

stoi = {ch:i for i,ch in enumerate(chars)}
itos = {i:ch for i,ch in enumerate(chars)}
encode = lambda s : [stoi[c] for c in s]  # str to int list
decode = lambda l : ''.join([itos[i] for i in l])  # int list to str

print(encode("hello"))
print(decode(encode("hello")))
```

```python
## tokenixe the original text and store it into a torch.Tensor

import torch
data = torch.tensor(encode(text),dtype=torch.long)
print(data.shape, data.dtype)
print(data[:100])   # corresponding tokenized text of first 1000 characters
```
